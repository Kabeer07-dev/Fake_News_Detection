```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
true_news = pd.read_csv(r"News_dataset\True.csv")
fake_news = pd.read_csv(r"News_dataset\Fake.csv")
```

```python
true_news['is_fake'] = 0
fake_news['is_fake'] = 1
```

```python
print(true_news.head())
print(fake_news.head())
```

```python
print("fake news: ",fake_news.shape)
print("\ntrue news: ",true_news.shape)
```

```python
df = pd.concat([true_news, fake_news], ignore_index=True)
```

```python
df = df.sample(frac=1, random_state=42).reset_index(drop=True)
```

```python
df = df.drop_duplicates()
```

```python
df.shape
```

```python
df['content'] = df['title'] + " " + df['text']
```

```python
df.head()
```

```python
df['content'] = df['content'].apply(lambda x:x.lower())
```

```python
df['content']
```

```python
## remove emojii

def remove_emoji(txt):
    new = " "
    for i in txt:
        if i.isascii():
            new+=i

    return new


df['content'] = df['content'].apply(remove_emoji)
```

```python
import string
def remove_punc(txt):
    return txt.translate(str.maketrans('','',string.punctuation))

df['content'] = df['content'].apply(remove_punc)
```

```python
df['content']
```

```python
import re

def remove_urls(text):
    text = re.sub(r'http\S+|www\S+|https\S+', '', text)
    return text


df['content'] = df['content'].apply(remove_urls)
```

```python
def remove_html(text):
    text = re.sub(r'<.*?>', '', text)
    return text

df['content'] = df['content'].apply(remove_html)
```
